## Step 1: Load .ubyte into NumPy (locally)

```python
import numpy as np

def load_images(path):
    with open(path, 'rb') as f:
        f.read(4)  # magic
        num = int.from_bytes(f.read(4), 'big')
        rows = int.from_bytes(f.read(4), 'big')
        cols = int.from_bytes(f.read(4), 'big')
        data = np.frombuffer(f.read(), dtype=np.uint8)
        return data.reshape(num, rows, cols)

def load_labels(path):
    with open(path, 'rb') as f:
        f.read(4)  # magic
        num = int.from_bytes(f.read(4), 'big')
        data = np.frombuffer(f.read(), dtype=np.uint8)
        return data
```

## Step 2: Convert to a structured dataset

```python
import pandas as pd

X_train = load_images("train-images.idx3-ubyte")
y_train = load_labels("train-labels.idx1-ubyte")

# flatten images
X_flat = X_train.reshape(X_train.shape[0], -1)

df = pd.DataFrame(X_flat)
df['label'] = y_train  # append labels at end

df.to_csv("mnist_train.csv", index=False)
print("Saved mnist_train.csv")
```

```python
X_test = load_images("t10k-images.idx3-ubyte")
y_test = load_labels("t10k-labels.idx1-ubyte")

X_test_flat = X_test.reshape(X_test.shape[0], -1)
df_test = pd.DataFrame(X_test_flat)
df_test['label'] = y_test

df_test.to_csv("mnist_test.csv", index=False)
```
